# 01 — Data Understanding (LEAD1.0-small)

Reproducible summary of the data audit. All numbers are **read from
`results/metrics/data_audit.json`** — nothing here is hardcoded.

Reproduce from scratch: `python scripts/run_all.py` (stage 1–2).

```python
import json
audit = json.load(open('../results/metrics/data_audit.json'))
v, q, t = audit['volume'], audit['quality'], audit['target']

print(f"rows            : {v['n_rows']:,}")
print(f"buildings       : {v['n_buildings']}")
print(f"span            : {v['timestamp_min']} .. {v['timestamp_max']} ({v['span_days']:.0f} days)")
print(f"missing reading : {q['missing_meter_reading']:,} ({q['missing_meter_reading_rate']:.2%})")
print(f"anomalies       : {t['n_anomaly']:,} ({t['anomaly_rate']:.2%}, imbalance 1:{t['class_imbalance_ratio']:.1f})")
```

```python
# Per-building anomaly distribution (from the audit, no recomputation)
import matplotlib.pyplot as plt

rates = [b['anomaly_rate'] for b in audit['per_building'].values()]
plt.figure(figsize=(8, 3))
plt.hist(rates, bins=30, color='#4472c4')
plt.xlabel('anomaly rate per building')
plt.ylabel('buildings')
plt.title('Anomaly rate distribution across 200 buildings')
plt.tight_layout(); plt.show()
```

**Takeaways:** 200 buildings × 8,784 hourly timestamps, 6.15% missing readings
(handled by D-009/D-011 — no imputation for the tree model), 2.13% anomalies.
Class imbalance ~1:46 drives the choice of PR-AUC as a co-primary metric (D-004).
